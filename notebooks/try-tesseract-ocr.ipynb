```python
import io

from pypdf import PdfReader

import pytesseract
from PIL import Image

# Path to the Tesseract-OCR executable

pytesseract.pytesseract.tesseract_cmd = "/opt/homebrew/bin/tesseract"
```

```python
def ocr_pdf(file_path):
    # Open the PDF file
    reader = PdfReader(file_path)
    text = {}

    # Iterate through each page
    for n_page, page in enumerate(reader.pages):
        # Check if the page contains images
        if '/XObject' in page['/Resources']:
            xObject = page['/Resources']['/XObject']

            for obj in xObject.keys():
                if xObject[obj]['/Subtype'] == '/Image':
                    size = (xObject[obj]['/Width'], xObject[obj]['/Height'])
                    data = xObject[obj].get_data()  # Use get_data() instead of _data
                    if xObject[obj]['/ColorSpace'] == '/DeviceRGB':
                        mode = "RGB"
                    else:
                        mode = "P"

                    # Perform OCR on the image data
                    image = Image.open(io.BytesIO(data))
                    text[n_page] = pytesseract.image_to_string(image, lang='eng')
    
    return text
```

```python

pdf_file_path = '/Users/mjboothaus/code/github/databooth/horse-logic/docs/provided/OCR/Exp2TimingSheets day 1.pdf'
```

```python
# Perform OCR on the PDF
extracted_text = ocr_pdf(pdf_file_path)
print(extracted_text)
```
