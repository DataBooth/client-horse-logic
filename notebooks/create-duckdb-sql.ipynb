```python
from dataclasses import fields
from datetime import datetime

import duckdb
from horse_logic.logfiles import Event, Experiment, Parameter, Trial
```

```python
def insert_record(con, record):
    table_name = record.__class__.__name__.lower() + 's'  # Assumes table name is plural of the class name
    columns = ', '.join(field.name for field in fields(record))
    placeholders = ', '.join('?' for _ in fields(record))
    values = tuple(getattr(record, field.name) for field in fields(record))
    query = f'INSERT INTO {table_name} ({columns}) VALUES ({placeholders})'
    con.execute(query, values)
```

```python
con = duckdb.connect(database='create-duckdb-sql.ddb')
```

```python
def create_tables_from_sql_file(con, file_path):
    with open(file_path, 'r') as file:
        sql_commands = file.read()
    con.execute(sql_commands)
```

```python
# Example usage
create_tables_from_sql_file(con, '../sql/create_tables_duckdb.sql')
```

```python
experiment = Experiment(1, 'Subject A', 1, 'Type 1', 'Initial test', datetime.now(), 'log1.txt')
insert_record(con, experiment)
```

```python
parameter = Parameter(1, 1, 'Param1', 'Value1')
insert_record(con, parameter)
```

```python
trial = Trial(1, 1, 1, datetime.now())
insert_record(con, trial)
```

```python
event = Event(1, 1, 'Start', datetime.now())
insert_record(con, event)
```

```python
con.close()
```
