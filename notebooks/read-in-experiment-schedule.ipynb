```python
import pandas as pd
import openpyxl
import os

# Define the path to your Excel file
excel_file_path = r'../docs/provided/wagga/EXP 2 design and Session structure_MJB_v2.xlsx'
sheet_name = 'TrialOrder'

# Define the range for the first table 
# usecols parameter takes Excel-style column letters or column index (0-indexed)
# skiprows skips rows before the start row (0-indexed)
# nrows reads a specific number of rows from the start row
```

```python

# Read the first table on the left
table_1_left = pd.read_excel(excel_file_path, sheet_name=sheet_name,
                             usecols='B:G', skiprows=11, nrows=10)

# Read the table below the first table on the left (B38:G47)
table_2_left = pd.read_excel(excel_file_path, sheet_name=sheet_name,
                             usecols='B:G', skiprows=27, nrows=10, header=None)

table_2_left.columns = table_1_left.columns

# Concatenate the two tables on the left to form a single DataFrame
df_left = pd.concat([table_1_left, table_2_left], ignore_index=True)

# Read the first table on the right
table_1_right = pd.read_excel(excel_file_path, sheet_name=sheet_name,
                              usecols='J:O', skiprows=11, nrows=10)

# Read the table below the first table on the right (J38:O47)
table_2_right = pd.read_excel(excel_file_path, sheet_name=sheet_name,
                              usecols='J:O', skiprows=27, nrows=10, header=None)

table_2_right.columns = table_1_right.columns

# Concatenate the two tables on the right to form a single DataFrame
df_right = pd.concat([table_1_right, table_2_right], ignore_index=True)
```

```python
# Reset the index to run from 1 to 20 for df_left
df_left.index = range(1, len(df_left) + 1)

# Ensure the column headings in df_right are identical to df_left
df_right.columns = df_left.columns

# Reset the index to run from 1 to 20 for df_right
df_right.index = range(1, len(df_right) + 1)
```

```python
# Ensure the original workbook exists
if not os.path.exists(excel_file_path):
    print(f"The file {excel_file_path} does not exist.")
else:
    # Load the original workbook
    book = openpyxl.load_workbook(excel_file_path)
    
    # If the 'Combined_Tables' sheet exists, remove it
    if 'Combined_Tables' in book.sheetnames:
        del book['Combined_Tables']
    book.save(excel_file_path)

    ws = book.create_sheet('Combined_Tables')

    rows = openpyxl.dataframe_to_rows(df_left, index=False)
    for r_idx, row in enumerate(rows, 1):
        for c_idx, value in enumerate(row, 1):
            ws.cell(row=r_idx, column=c_idx, value=value)
```
