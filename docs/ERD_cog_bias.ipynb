## ERD - Cognitive Bias Experiments

```mermaid
erDiagram
    ExperimentCB ||--o{ TrialCB : contains
    TrialCB ||--o{ EventCB : has
    TrialCB ||--o| ResponseCB : may_have

    ExperimentCB {
        int ExperimentID PK
        string Cohort
        string SubjectName
        int SubjectNumber
        int SessionNumber
        string SessionType
        string ExperimentType
        string Comment
        datetime DateTime
        string LogFileName
        json Parameters
    }

    TrialCB {
        int TrialID PK
        int ExperimentID FK
        int TrialNumber
        string ResponseType
        string TrialSubType
        string Direction
        string CriterionType
        float CriterionCount
        datetime TrialStartTime
        datetime TrialEndTime
        string SessionType
    }

    ResponseCB {
        int ResponseID PK
        int TrialID FK
        float ResponseTime
        datetime EventDateTime
    }

    EventCB {
        int EventID PK
        int TrialID FK
        string EventType
        datetime EventTime
        float ElapsedTime
    }
```

### Propose changes

The updated ERD diagram above includes the following changes:

- Added `SessionType` to the `ExperimentCB` entity.
- Added `SessionType` to the `TrialCB` entity.
- Kept the relationships between entities the same.

The diagram shows:
- An `ExperimentCB` can have multiple `TrialCB`s (one-to-many relationship).
- A `TrialCB` can have multiple `EventCB`s (one-to-many relationship).
- A `TrialCB` may have one `ResponseCB` (one-to-one relationship, but it's optional as indicated by the circle on the Trial side).
- The `SessionType` field in both `ExperimentCB` and `TrialCB` allows for distinguishing between "training" and "test" sessions at both the experiment and trial levels.

This structure hopefully provides flexibility in your data model and allows for easy querying of session types at different levels of granularity. However, you may want to only have it stored in one table.
